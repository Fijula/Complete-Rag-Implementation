```python
from openai import OpenAI
client = OpenAI()
response= client.embeddings.create(model="text-embedding-3-small",input=text)
openai_embeddings= response.data[0].embedding
print("Provider : OpenAI")
print("Dimensions",len(openai_embeddings))
print("Embeddings",openai_embeddings)
```

```python
from sentence_transformers import SentenceTransformer
model =SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
hf_embeddings = model.encode(text).tolist()
print("Provider : HuggingFace")
print("Dimensions",len(hf_embeddings))
print("Embeddings",hf_embeddings)
```
